```python
# -- Importar as bibliotecas
import numpy as np
import matplotlib.pyplot as plt
```

### Arrays

Um *array* é uma coleção de números. Podemos criar arrays de diferentes maneiras.

```python
array1 = np.array([1, 2, 3, 4, 5])
array2 = np.arange(1,6)

print('Estes dois arrays são iguais!')
print(array1)
print(array2)

array3 = np.arange(0,6,0.5)
print('\nEste array tem um espaçamento menor entre os valores')
print(array3)
```

## Operações matemáticas com arrays

As operações matemáticas básicas com arrays são realizadas elemento a elemento.

```python
x = array1
y =2*x

print("Acabei de multiplicar o array por dois!")
print('x:', x)
print('y:', y)

print("Também posso somar dois arrays")
z = x + y
print('z = x+y:', z)
```

## Gráficos com arrays

Podemos usar o `pyplot` para construir gráficos a partir de arrays.

```python
plt.plot(x,y)
```

```python
# -- Mesmo gráfico, mas com mais informações
plt.plot(x,y)
plt.title("Título do gráfico")
plt.xlabel('Valores de x')
plt.ylabel('Valores de y')
```
