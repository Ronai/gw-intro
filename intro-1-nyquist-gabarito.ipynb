# Gabarito — Frequência de amostragem e frequência de Nyquist

Resolução das **questões de verificação** do notebook `intro-1-nyquist.ipynb` (versão em português brasileiro).

Neste exemplo, o sinal é amostrado com frequência $f_s=20\,\mathrm{Hz}$, durante $1\,\mathrm{s}$.

> **Objetivo:** distinguir frequência de amostragem, intervalo entre amostras, frequência de Nyquist e número de amostras por ciclo; compreender por que uma senoide de 10 Hz pode não ser observada nas amostras.

## Questões de verificação

1. Qual é a frequência de amostragem do vetor de tempo criado na primeira célula de código?
2. Quantas amostras existem em 1 segundo de dados?
3. Qual é a frequência de Nyquist para essa frequência de amostragem?
4. Quantas amostras por ciclo existem para o sinal de 5 Hz?
5. Por que o sinal de 10 Hz desaparece? Você consegue explicar esse comportamento?

## Respostas às questões 1–4

**1. Frequência de amostragem:**

$$\boxed{f_s=20\,\mathrm{Hz}}$$

Isso significa que são coletadas **20 amostras por segundo**. O intervalo entre amostras é

$$\Delta t=\frac{1}{f_s}=\frac{1}{20}=0{,}05\,\mathrm{s}. $$

**2. Número de amostras em 1 segundo:**

$$\boxed{N=f_s T=20\times 1=20\text{ amostras}.}$$

Os instantes registrados são $0$, $0{,}05$, $0{,}10$, $\ldots$, $0{,}95\,\mathrm{s}$. Embora o último instante seja $0{,}95\,\mathrm{s}$, as 20 amostras representam um intervalo de aquisição de $1\,\mathrm{s}$.

**3. Frequência de Nyquist:**

$$\boxed{f_N=\frac{f_s}{2}=10\,\mathrm{Hz}.}$$

**4. Amostras por ciclo de um sinal de 5 Hz:**

$$\boxed{\frac{f_s}{f}=\frac{20}{5}=4\text{ amostras por ciclo}.}$$

O período do sinal de 5 Hz é $T_5=1/5=0{,}2\,\mathrm{s}$ e, portanto, $T_5/\Delta t=0{,}2/0{,}05=4$.

```python
# Conferir os resultados com os mesmos parâmetros do notebook original
import numpy as np
import matplotlib.pyplot as plt

fs = 20.0
time = np.arange(0, 1.0, 1.0 / fs)
f_sinal = 5.0

print(f"1. Frequência de amostragem: {fs:g} Hz")
print(f"2. Número de amostras: {len(time)}")
print(f"3. Frequência de Nyquist: {fs/2:g} Hz")
print(f"4. Amostras por ciclo de 5 Hz: {fs/f_sinal:g}")
print(f"   Intervalo entre amostras: {1/fs:.2f} s")
print(f"   Último instante registrado: {time[-1]:.2f} s")
```

## Questão 5 — Por que o sinal de 10 Hz desaparece?

A função amostrada no notebook é

$$y(t)=\sin(2\pi f t).$$

Para $f=10\,\mathrm{Hz}$, com frequência de amostragem $f_s=20\,\mathrm{Hz}$, os instantes de aquisição são

$$t_n=n\Delta t=\frac{n}{20},\qquad n=0,1,\ldots,19.$$

Substituindo esses instantes na função:

$$
\begin{aligned}
y(t_n)&=\sin\left(2\pi\cdot 10\cdot\frac{n}{20}\right)\\
&=\sin(n\pi).
\end{aligned}
$$

Como $\sin(n\pi)=0$ para qualquer inteiro $n$, resulta que

$$\boxed{y(t_n)=0\quad\text{para todas as amostras}.}$$

**Interpretação:** a senoide não desapareceu fisicamente. O que ocorre é que a amostragem foi realizada exatamente nos instantes em que ela cruza o zero. A frequência de 10 Hz corresponde à **frequência de Nyquist** para $f_s=20\,\mathrm{Hz}$, isto é, apenas duas amostras por ciclo, o que não permite recuperar de maneira inequívoca uma senoide de fase desconhecida.

```python
# Comparar a senoide de 10 Hz com os pontos amostrados a 20 Hz
f = 10.0
fs_alta = 1000.0

t_continuo = np.arange(0, 0.5, 1 / fs_alta)
y_continuo = np.sin(2 * np.pi * f * t_continuo)

t_amostras = np.arange(0, 0.5, 1 / fs)
y_amostras = np.sin(2 * np.pi * f * t_amostras)

plt.figure(figsize=(10, 4))
plt.plot(t_continuo, y_continuo, label='Sinal de 10 Hz (referência de alta resolução)')
plt.plot(t_amostras, y_amostras, 'o', markersize=7,
         label='Amostras coletadas a 20 Hz')
plt.axhline(0, color='gray', linewidth=0.8)
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.title('Amostras nos zeros da senoide de 10 Hz')
plt.ylim(-1.15, 1.15)
plt.grid(alpha=0.3)
plt.legend()
plt.show()

# Por erros numéricos de arredondamento, os valores podem não ser exatamente zero.
print('Amostras numericamente próximas de zero?', np.allclose(y_amostras, 0))
print('Maior amplitude absoluta registrada:', np.max(np.abs(y_amostras)))
```

### E se a fase do sinal for diferente?

Considere agora o cosseno com a mesma frequência:

$$x(t)=\cos(2\pi\cdot 10 t).$$

Nos mesmos instantes de amostragem,

$$x(t_n)=\cos(n\pi)=(-1)^n.$$

Portanto, as amostras alternam entre **+1 e −1**, em vez de desaparecerem. Isso mostra que o resultado depende da **fase** da onda em relação aos instantes de amostragem.

```python
# Mesma frequência, mesma amostragem: seno e cosseno têm resultados diferentes
f = 10.0
y_seno = np.sin(2 * np.pi * f * t_amostras)
y_cosseno = np.cos(2 * np.pi * f * t_amostras)

print('Seno (valores arredondados):', np.round(y_seno[:8], 10))
print('Cosseno:', np.round(y_cosseno[:8], 10))

plt.figure(figsize=(10, 4))
plt.plot(t_amostras, y_seno, 'o-', label='Seno de 10 Hz')
plt.plot(t_amostras, y_cosseno, 's-', label='Cosseno de 10 Hz')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude amostrada')
plt.title('Amostragem a 20 Hz: dependência da fase')
plt.ylim(-1.2, 1.2)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## Conclusão

No exemplo, $f_s=20\,\mathrm{Hz}$ e $f_N=10\,\mathrm{Hz}$. Uma senoide de 10 Hz, sem deslocamento de fase, produz apenas amostras nulas porque $\sin(n\pi)=0$. Já o cosseno de 10 Hz gera uma sequência alternada de valores $+1$ e $-1$.

Isso ilustra a importância de usar, para sinais limitados a uma frequência máxima $f_{\max}$, uma frequência de amostragem **estritamente maior** que o dobro dessa frequência:

$$\boxed{f_s>2 f_{\max}.}$$
