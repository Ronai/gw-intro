## Questões propostas

1. Defina um array `x` de 0 a 5, com espaçamento de 0,1 entre os pontos.
2. Defina um array `y` com o quadrado de cada valor de `x` (dica: `x**2`).
3. Faça o gráfico de `y` em função de `x`. Você consegue identificar uma parábola?

## 1. Construir o array `x`

Usamos `np.arange(início, fim, passo)`. Como o limite superior não é incluído, escolhemos **5,1** para que o valor **5,0** esteja no array.

O intervalo de 0 a 5, inclusive, com passo 0,1 contém $N=5/0{,}1+1=51$ pontos.

```python
# -- Importar as bibliotecas
import numpy as np
import matplotlib.pyplot as plt

# -- Criar o array de 0 a 5, com passo de 0,1
x = np.arange(0, 5.1, 0.1)

print("Array x:", x)
print("Número de pontos:", len(x))
print("Primeiro valor:", x[0])
print("Último valor:", round(x[-1], 1))
```

## 2. Calcular `y = x**2`

A operação `x**2` eleva **cada elemento do array** ao quadrado, sem a necessidade de utilizar um laço de repetição.

Matematicamente, $y=x^2$.

```python
# -- Elevar os elementos de x ao quadrado
y = x**2

print("Array y:", y)
print("Primeiros cinco pares (x, y):")
for xi, yi in zip(x[:5], y[:5]):
    print(f"x = {xi:.1f}  →  y = {yi:.2f}")
```

## 3. Representar `y` em função de `x`

O gráfico de $y=x^2$ é uma **parábola** com concavidade voltada para cima e vértice na origem $(0,0)$.

**Atenção:** como o exercício utiliza apenas $0\leq x\leq 5$, o gráfico apresenta **somente o ramo direito** da parábola.

```python
# -- Fazer o gráfico de y em função de x
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=r"$y = x^2$")
plt.title("Gráfico de uma função quadrática")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## Conclusão

O array `x` contém **51 pontos** igualmente espaçados de **0,1** entre **0** e **5**. O array `y` é obtido aplicando `x**2` a todos os pontos. O gráfico mostra o ramo direito da parábola $y=x^2$.

*Observação:* para ver os dois ramos da parábola, seria necessário escolher também valores negativos de `x`.
